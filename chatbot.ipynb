```python
!pip install -q -U google-genai
from google import genai
from google.colab import userdata
client = genai.Client(api_key=userdata.get('GOOGLE_API_KEY'))
```

```python
print("---시스템 체크 시작---")

#에러 없이 모델 목록 가져오기
available_models = [m.name for m in client.models.list()]
print("현재 사용 가능한 모든 모델; (available_models)")

#'gemini-1.5-flash'가 있으면 쓰고, 없으면 목록의 첫 번째 모델을 선택
target_model = 'model/gemini-flash-lite-latest'
if 'models/gemini-flash-lite-latest' in available_models:
  target_model = 'models/gemini-flash-lite-latest'
elif len(available_models) > 0:
  target_model = available_models[0]

print(f"사용할 모델: {target_model}")

#즉시 테스트 호출
try:
  response = client.models.generate_content(
      model = target_model,
      content = "정상 작동 중인가요?"
  )
  print(f"\n[성공] AI: 응답: {response.text}")
except Exception as e:
  print(f"\n[실패] 에러 발생: {e}")
  print("Tip: API Studiodptj 'Create API key in new project'로 새 키를 발급받아보세요.")
```

```python
#페르소나 설정 및 대화 시작
sys_instruct = """
당신은 미술관 큐레이터입니다
1. 모든 답변은 예술 지식을 기반으로 전문적으로 답변하세요.
2. 답변에 오류가 없는지 확인한 후 답변하세요.
3. 설명에 대한 근거로는 항상 출처를 함께 제시하세요.
"""

#대화 세션 생성
chat = client.chats.create(
  model = target_model,
  config = {'system_instruction': sys_instruct}

)
```

```python
print("---챗봇과 대화를 시작합니다!(종료: quit)---")

while True:
  user_input = input("학생:")
  if user_input.lower() in ['quit','exit','종료']:
    print("대화를 종료합니다. 수고하셨습니다!")
    break

#메세지 전송 및 응답 출력
response = chat.send_message(user_input)
print(f"봇: {response.text}")
```
